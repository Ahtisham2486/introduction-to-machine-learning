Module : a module is a single file conaining the definitions and the statements


> Package : it is the collection of the python modules under the comman namespace , this is achieved by having a directory with special .__init__ file


> library : it is the collection of the packages or the modules that are grouped together to provide the functionality that can be utilized by the other programms






```python
class BaseSalary:
  def __init__(self,BaseSalary, bonous_rate=0.1,symbol="$" ):
    self.BaseSalary= BaseSalary
    self.bonous_rate=bonous_rate
    self.symbol=symbol
  def __repr__(self):
    return f"{self.symbol}{self.BaseSalary:,.0f}"


  def calaculate_total_salary(self):
    total_Salary= self.BaseSalary * (1+ self.bonous_rate)
    return total_Salary
```
